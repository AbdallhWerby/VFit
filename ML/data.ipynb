```python
import random
import csv

def generate_vitals_data(duration):
    skin_temp = []
    heart_rate = []
    spo2 = []
    target = []
    
    def generate_vitals_params(stage):
        if stage == 'sprint':
            temp_range = (36.6, 37.2)
            rate_range = (120, 150)
            spo2_range = (90, 95)
            target_val = 'Stressed'
        elif stage == 'alert':
            temp_range = (38.0, 41)
            rate_range = (160, 220)
            spo2_range = (85, 93)
            target_val = 'Alert'
        else:
            temp_range = (35, 36)
            rate_range = (80, 120)
            spo2_range = (95, 100)
            target_val = 'Normal'
        
        temp = random.uniform(*temp_range)
        rate = random.randint(*rate_range)
        spo2_value = random.randint(*spo2_range)
        
        return round(temp, 1), rate, spo2_value, target_val
    
    def generate_sprint_data():
        for i in range(60):
            if i < 10:
                temp, rate, spo2_value, target_val = generate_vitals_params('normal')
            elif i < 45:
                temp, rate, spo2_value, target_val = generate_vitals_params('sprint')
            else:
                temp, rate, spo2_value, target_val = generate_vitals_params('alert')
            
            skin_temp.append(temp)
            heart_rate.append(rate)
            spo2.append(spo2_value)
            target.append(target_val)
    
    def generate_break_data():
        for i in range(30):
            temp, rate, spo2_value, target_val = generate_vitals_params('normal')
            skin_temp.append(temp)
            heart_rate.append(rate)
            spo2.append(spo2_value)
            target.append(target_val)
    
    # Generate initial sprint
    generate_sprint_data()
    
    # Generate break period
    generate_break_data()
    
    # Generate additional sprints until 30 minutes
    while len(skin_temp) < duration:
        generate_sprint_data()
    
    # Create a CSV file to save the generated data
    with open('vitals_datareal.csv', 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Skin_temp', 'Heart_rate', 'SPO2', 'Target'])
        for i in range(duration):
            writer.writerow([skin_temp[i], heart_rate[i], spo2[i], target[i]])

# Call the function to generate 30 minutes (1800 seconds) of data
generate_vitals_data(1800)
```

```python
import random
import csv

def generate_vitals_data(duration):
    skin_temp = []
    heart_rate = []
    spo2 = []
    target = []
    row = []
    
    def generate_normal_data(start_row):
        start_temp = random.uniform(34, 36)
        start_rate = random.randint(80, 120)
        start_spo2 = random.randint(96, 100)
        
        temp_range = (0.1, 0.5)
        rate_range = (1, 3)
        spo2_range = (0.1, 0.5)
        
        temp = start_temp
        rate = start_rate
        spo2_value = start_spo2
        
        for i in range(60):
            skin_temp.append(round(temp, 1))
            heart_rate.append(rate)
            spo2.append(spo2_value)
            target.append('Normal')
            row.append(start_row + i)
            
            temp_increment = random.uniform(*temp_range)
            rate_increment = random.randint(*rate_range)
            spo2_decrement = random.uniform(*spo2_range)
            
            temp = min(max(temp + temp_increment, 34), 36)
            rate = min(max(rate + rate_increment, 80), 120)
            spo2_value = min(max(spo2_value - spo2_decrement, 96), 100)
    
    def generate_stressed_data(start_row):
        start_temp = random.uniform(36.6, 37.2)
        start_rate = random.randint(120, 150)
        start_spo2 = random.uniform(90, 95)
        
        temp_range = (0.1, 0.5)
        rate_range = (1, 3)
        spo2_range = (0.1, 0.5)
        
        temp = start_temp
        rate = start_rate
        spo2_value = start_spo2
        
        for i in range(60):
            skin_temp.append(round(temp, 1))
            heart_rate.append(rate)
            spo2.append(spo2_value)
            target.append('Stressed')
            row.append(start_row + i)
            
            temp_increment = random.uniform(*temp_range)
            rate_increment = random.randint(*rate_range)
            spo2_decrement = random.uniform(*spo2_range)
            
            temp = min(max(temp + temp_increment, 36.6), 37.2)
            rate = min(max(rate + rate_increment, 120), 150)
            spo2_value = min(max(spo2_value - spo2_decrement, 90), 95)
    
    def generate_alert_data(start_row):
        start_temp = random.uniform(38.0, 41)
        start_rate = random.randint(160, 220)
        start_spo2 = random.uniform(85, 93)
        
        temp_range = (0.1, 0.5)
        
        temp = start_temp
        
        for i in range(30):
            skin_temp.append(round(start_temp, 1))
            heart_rate.append(start_rate)
            spo2.append(start_spo2)
            target.append('Alert')
            row.append(start_row + i)
            
            temp_decrement = random.uniform(*temp_range)
            
            start_temp = min(max(start_temp - temp_decrement, 38.0), 41)
    
    # Generate initial normal data
    generate_normal_data(1)
    
    # Generate stressed data
    generate_stressed_data(61)
    
    # Generate alert data
    generate_alert_data(121)
    
    # Generate additional data until reaching the duration
    i = 181
    while len(skin_temp) < duration:
        generate_normal_data(i)
        i += 60
        generate_stressed_data(i)
        i += 60
        generate_alert_data(i)
        i += 30
    
    # Create a CSV file to save the generated data
    with open('vitals_datatimeu.csv', 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Row', 'Skin_temp', 'Heart_rate', 'SPO2', 'Target'])
        for i in range(duration):
            writer.writerow([row[i], skin_temp[i], heart_rate[i], spo2[i], target[i]])

# Call the function to generate 30 minutes (1800 seconds) of data
generate_vitals_data(1800)
```
